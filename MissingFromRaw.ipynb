```python
import numpy as np
import pandas as pd
from scipy.stats import zscore
import matplotlib.pyplot as plt

# df = pd.DataFrame(data, columns=['Value'])

data = pd.read_csv('https://raw.githubusercontent.com/raghurammudumbi-collab/MachineLearningUsingPython/refs/heads/main/JSCD-TEACHERS.csv')

df = pd.DataFrame(data, columns=['TSAL'])

df['Z-score'] = zscore(df['TSAL'])
print(df)

outliers = df[df['Z-score'].abs() > 2]
print(outliers)
```

```python
import numpy as np
import pandas as pd
from scipy.stats import zscore
import matplotlib.pyplot as plt

# df = pd.DataFrame(data, columns=['Value'])

data = pd.read_csv('https://raw.githubusercontent.com/raghurammudumbi-collab/MachineLearningUsingPython/refs/heads/main/JSCD-STUDENTS.csv')

df = pd.DataFrame(data, columns=['SMARKS'])
print(df)
df['Z-score'] = zscore(df['SMARKS'])
print(df)

outliers = df[df['Z-score'].abs() > 2]
print(outliers)
```
