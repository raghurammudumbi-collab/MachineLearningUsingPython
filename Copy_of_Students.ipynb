```python
import numpy as np
import pandas as pd
df = pd.read_csv("https://raw.githubusercontent.com/raghurammudumbi-collab/MachineLearningUsingPython/refs/heads/main/JSCD-STUDENTS.csv")
data = df['SMARKS']
# Calculate Population Variance (Default: ddof=0)
pop_var = np.var(data)
print(f"NumPy Population Variance: {pop_var}")  # Output: 2.0
```
