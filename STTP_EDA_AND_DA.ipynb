```python
# Histogram

import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
data = pd.read_csv('https://raw.githubusercontent.com/raghurammudumbi-collab/MachineLearningUsingPython/refs/heads/main/JSCD-TEACHERS.csv')
print(data.head())
```

```python
sns.histplot(data['TAGE'])
```

```python
x = data['TMRF'].value_counts()
plt.pie(x.values,
        labels=x.index,
        autopct='%1.1f%%')
plt.show()
```

```python
import matplotlib.pyplot as plt
plt.figure(figsize=(15, 5))
sns.barplot(x=data['TNAME'], y=data['TSAL'])
plt.xticks(rotation=90)
```

```python
sns.scatterplot(x=data['TAGE'],
                y=data['TSAL'])
```
