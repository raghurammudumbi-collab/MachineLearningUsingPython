```python
def findMean(a, n):
    sum = 0
    for i in range(0, n):
        sum += a[i]
    return float(sum / n)
# Driver program
a = [1, 3, 4, 2, 7, 5, 8, 6]
n = len(a)
print("Mean =", findMean(a, n))
```

```python
import statistics
data = [1, 2, 3, 4, 5]
# Calculate Sample Variance (Denominator is n - 1)
sample_var = statistics.variance(data)
print(f"Sample Variance: {sample_var}")  # Output: 2.5
# Calculate Population Variance (Denominator is n)
pop_var = statistics.pvariance(data)
print(f"Population Variance: {pop_var}")  # Output: 2.0
```

```python
import numpy as np
data = [1, 2, 3, 4, 5]
# Calculate Population Variance (Default: ddof=0)
pop_var = np.var(data)
print(f"NumPy Population Variance: {pop_var}")  # Output: 2.0
# Calculate Sample Variance (Set Delta Degrees of FrAdidtribution in statistics is a mathematical function, table, or graph that describes how the values of a variable are spread out across a dataset. It shows which values occur frequently, which are rare, and whether the data is clustered, symmetric, or skewed. [1, 2, 3]eedom to 1)
sample_var = np.var(data, ddof=1)
print(f"NumPy Sample Variance: {sample_var}")  # Output: 2.5
```

```python
import numpy as np
import matplotlib.pyplot as plt

# Creating a series of data of in range of 1-50.
x = np.linspace(1,50,200)

#Creating a Function.
def normal_dist(x , mean , sd):
    prob_density = (np.pi*sd) * np.exp(-0.5*((x-mean)/sd)**2)
    return prob_density

#Calculate mean and Standard deviation.
mean = np.mean(x)
sd = np.std(x)

#Apply function to the data.
pdf = normal_dist(x,mean,sd)

#Plotting the Results
plt.plot(x,pdf , color = 'red')
plt.xlabel('Data points')
plt.ylabel('Probability Density')
```

```python

import numpy as np
import scipy.stats as stats
from statsmodels.stats.weightstats import ztest

# Set random seed for reproducibility
np.random.seed(42)
alpha = 0.05
data = np.random.normal(loc=50, scale=5, size=100)

stat, p_val = stats.shapiro(data)
print(f"Shapiro P-value: {p_val:.4f}")
# Output decision rule
if p_val <= alpha:
    print("Reject H0: Data is NOT normally distributed.")
else:
    print("Fail to reject H0: Data appears normally distributed.")
```
